```python
# program 3

numgrade = int(input("Enter the numerical grade: "))

if numgrade <= 100 and numgrade >= 90:
  print("A")
elif numgrade <= 89 and numgrade >= 80:
  print("B")
elif numgrade <= 79 and numgrade >=70:
  print("C")
elif numgrade < 70:
  print("F")
```
